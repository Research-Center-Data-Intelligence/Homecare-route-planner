```python
# make a csv with the following columns:
# - name
# - description
# Make 4 rows with the following data:
# - name: "Alice", description: "A software engineer with 5 years of experience."
# - name: "Bob", description: "A data scientist who loves machine learning."

import csv

data = [
    {"name": "Alice", "description": "A software engineer with 5 years of experience."},
    {"name": "Bob", "description": "A data scientist who loves machine learning."},
    {"name": "Charlie", "description": "A product manager with a passion for user experience."},
    {"name": "Diana", "description": "A UX designer who enjoys creating intuitive interfaces."}
]

fieldnames = ["name", "description"]

with open("people.csv", "w", newline="") as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for row in data:
        writer.writerow(row)

print("CSV file 'people.csv' has been created with the specified data.")
```
