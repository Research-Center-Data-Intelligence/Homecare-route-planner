```python
import csv

data = [
    {
        "naam": "Client 1",
        "adres": "Kerkstraat 12 6411AA Heerlen",
        "zorgarrangement": "HBH Basis",
        "voorkeuren": "niet op dinsdag",
        "overige_info": "heeft hond",
        "aantal_uren_zorg": "2"
    },
]

fieldnames = [
    "naam",
    "adres",
    "zorgarrangement",
    "voorkeuren",
    "overige_info",
    "aantal_uren_zorg"
]

with open("clients.csv", "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for row in data:
        writer.writerow(row)

print("CSV file 'clients.csv' has been created with the specified data.")
```
